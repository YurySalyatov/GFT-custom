```python
import itertools
import os

all_datasets = ['aminer', 'arnetminer', 'kisti', 'pbmed', 'qian', 'zbmath']
# all_datasets = ['arnetminer', 'pbmed', 'qian', 'zbmath']
# Генерируем все непустые подмножества (2^6 - 1 = 63)
dataset_sets = []
for r in range(1, len(all_datasets) + 1):
    for combo in itertools.combinations(all_datasets, r):
        dataset_sets.append(list(combo))

print(f"Всего наборов: {len(dataset_sets)}")
# Для отладки выведем первые 5
for i, s in enumerate(dataset_sets[:5]):
    print(f"{i+1}: {s}")
# dataset_sets = dataset_sets[:5]
# dataset_sets = dataset_sets[5:10]
dataset_sets = dataset_sets[10:15]
```

```python
import os
import shutil
import time

base_ckpt = 'ckpts/pretrain_model'
embeddings_root = 'embeddings'
epoch_to_load = 50  # эпоха, которую будем использовать для извлечения (можно менять)

# Создаём корневые папки
os.makedirs(base_ckpt, exist_ok=True)
os.makedirs(embeddings_root, exist_ok=True)

for idx, dataset_list in enumerate(dataset_sets):
    set_name = '_'.join(dataset_list)  # например "arnetminer_qian"
    print(f"\n=== Обработка набора {idx+1}/{len(dataset_sets)}: {set_name} ===")
    start_time = time.time()

    # 1. Обновляем конфиг для обучения
    !python change_yaml_config.py --datasets {','.join(dataset_list)} --config config/pt_data.yaml --default_weight 1.0

    # 2. Запускаем предобучение с сохранением в отдельную папку
    save_dir = f"{base_ckpt}/"
    !mkdir -p {save_dir}
    !python GFT/pretrain.py --use_params --debug --save_dir {save_dir} --save_name {set_name}
    
    checkpoint_folder = f"{base_ckpt}/codebook_size_128_layer_2_pretrain_on_{set_name}_seed_42"
    dst_model = f"{embeddings_root}/{set_name}/codebook_size_128_layer_2_pretrain_on_{set_name}_seed_42"
    encoder = f"{checkpoint_folder}/encoder_{epoch_to_load}.pt"
    vq = f"{checkpoint_folder}/vq_{epoch_to_load}.pt"
    os.makedirs(dst_model, exist_ok=True)
    shutil.move(encoder, f"{dst_model}/encoder_{epoch_to_load}.pt")
    shutil.move(vq, f"{dst_model}/vq_{epoch_to_load}.pt")
    info_file = f"{checkpoint_folder}/info.txt"
    if os.path.exists(info_file):
        shutil.move(info_file, f"{dst_model}/info.txt")
    shutil.rmtree(checkpoint_folder)
    # 3. Извлекаем эмбеддинги для ВСЕХ датасетов (из all_datasets)
    for ds in all_datasets:
        print(f"  Извлечение для {ds}...")
        out_dir = f"{embeddings_root}/{set_name}/{ds}"
        !python GFT/extract_embs.py --dataset {ds} --checkpoint_dir {dst_model} --epoch {epoch_to_load} --output_dir {out_dir}

    elapsed = time.time() - start_time
    print(f"Набор {set_name} завершён за {elapsed:.2f} сек.")

print("Все наборы обработаны.")
```

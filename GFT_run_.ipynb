```python
import yaml
# ===== Загружаем конфиг pretrain.yaml для получения параметров модели =====
with open('config/pretrain.yaml', 'r') as f:
    pretrain_config = yaml.safe_load(f)

# Рекурсивный поиск значения по ключу в словаре (разной вложенности)
def find_param(cfg, key):
    if key in cfg:
        return cfg[key]
    for v in cfg.values():
        if isinstance(v, dict):
            val = find_param(v, key)
            if val is not None:
                return val
    return None

codebook_size = find_param(pretrain_config, 'codebook_size')
num_layers = find_param(pretrain_config, 'num_layers')
if num_layers is None:
    num_layers = find_param(pretrain_config, 'layers')

if None in (codebook_size, num_layers):
    raise ValueError("Не удалось найти все необходимые параметры (codebook_size, num_layers/layers, seed) в config/pretrain.yaml")
print(f"codebook size:{codebook_size}\nlayers:{num_layers}")
```
